# CompleteJourneyExtractor の実行・確認

`CompleteJourneyExtractor` で Complete Journey の各テーブルを抽出し、
内容（形状・型・先頭行）を確認するためのノートブック。

```python
from pathlib import Path

import pandas as pd

from ariadne.etl.completejourney.extract import (
    CompleteJourneyExtractor,
    DEFAULT_TABLE_TYPES,
)

# リポジトリルート（notebooks/005_.../ から 2 階上）
PROJECT_ROOT = Path.cwd().parents[1]
print("PROJECT_ROOT:", PROJECT_ROOT)
print("raw_dir exists:", (PROJECT_ROOT / "data/00_raw/completejourney/rdata").exists())
```

## 単一テーブルの抽出

```python
extractor = CompleteJourneyExtractor(PROJECT_ROOT)
# extract_all() のデフォルト引数はモジュール定義時の DEFAULT_TABLE_TYPES に束縛されるため、
# 読み込みたいテーブルは明示的に渡す必要がある。
tables = extractor.extract_all(
    table_types = {
        'campaign_descriptions': 'rda'
        }
    )
sorted(tables.keys())
```

## 全テーブルの抽出

```python
extractor = CompleteJourneyExtractor(PROJECT_ROOT)
# extract_all() のデフォルト引数はモジュール定義時の DEFAULT_TABLE_TYPES に束縛されるため、
# 読み込みたいテーブルは明示的に渡す必要がある。
tables = extractor.extract_all(
sorted(tables.keys())
```

## 各テーブルの概要（型・形状）

```python
for name in sorted(tables):
    obj = tables[name]
    kind = type(obj).__name__
    shape = getattr(obj, "shape", None)
    print(f"{name:25s} type={kind:12s} shape={shape}")
```

## 個別テーブルの中身確認

`table_name` を変えて各テーブルの `dtypes` と先頭行を確認できます。

```python
table_name = "transactions"
df = tables[table_name]
print(type(df))
if isinstance(df, pd.DataFrame):
    display(df.dtypes)
    display(df.head())
```
