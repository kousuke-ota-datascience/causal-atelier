因果探索から因果推論までの実行計画と成果物の受け渡しを確認する。

```python
import sys
from pathlib import Path

cwd = Path.cwd().resolve()
for candidate in (cwd, *cwd.parents):
    if (candidate / 'pyproject.toml').exists():
        PROJECT_ROOT = candidate
        break
else:
    raise RuntimeError('repository root was not found')

sys.path.insert(0, str(PROJECT_ROOT / 'src'))
```

```python
from ariadne.application.pipeline.planning import PipelinePlanner
from ariadne.interfaces.cli.pipeline import parse_args

args = parse_args([
    '--project-root', str(PROJECT_ROOT),
    '--pipeline-config', str(PROJECT_ROOT / 'configs/causal/inference/pipeline.yaml'),
    '--dry-run',
])
plan = PipelinePlanner(PROJECT_ROOT).build_plan(args, strategy_name='dry_run')
plan_dict= plan.to_dict()
```
